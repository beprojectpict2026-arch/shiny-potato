**Problem Statement:** Implement basic logic gates using Hebbnet neural networks

**Dataset:** Truth tables of the AND, OR, NAND and NOR gates (and XOR for comparison) in bipolar form (1 = true, -1 = false)

**Requirements:** Python 3

**Code:-**

```python
inputs = [(1, 1), (1, -1), (-1, 1), (-1, -1)]
gates = {
    "AND": [1, -1, -1, -1],
    "OR": [1, 1, 1, -1],
    "NAND": [-1, 1, 1, 1],
    "NOR": [-1, -1, -1, 1],
    "XOR": [-1, 1, 1, -1],
}
```

```python
def train_hebb(inputs, targets):
    w, b = [0, 0], 0
    for x, t in zip(inputs, targets):
        w = [wi + xi * t for wi, xi in zip(w, x)]
        b += t
    return w, b


def activation(net):
    if net > 0:
        return 1
    if net < 0:
        return -1
    return 0
```

```python
for name, targets in gates.items():
    w, b = train_hebb(inputs, targets)
    print(f"{name} gate: weights = {w}, bias = {b}")
    print("  x1  x2 | target | net | output")
    correct = 0
    for x, t in zip(inputs, targets):
        net = b + sum(wi * xi for wi, xi in zip(w, x))
        out = activation(net)
        correct += out == t
        print(f"  {x[0]:>2}  {x[1]:>2} | {t:>5}  | {net:>3} | {out:>5}")
    print(f"  Rows correct: {correct}/4\n")
```
